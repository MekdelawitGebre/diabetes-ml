```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path().resolve().parent  # adjust if notebook is in a subfolder
sys.path.append(str(project_root))


import matplotlib.pyplot as plt
import seaborn as sns
from src.data.preprocess import load_data, clean_data
from src.data.feature_engineering import add_age_bins, add_bmi_features

df = load_data()
df = clean_data(df)
df = add_age_bins(df)
df = add_bmi_features(df)

# Glucose distribution
sns.histplot(df["Glucose"], bins=30, kde=True)
plt.title("Glucose Distribution")
plt.savefig("plots/glucose_distribution.png")
plt.close()

# Outcome counts
sns.countplot(x="Outcome", data=df)
plt.title("Diabetes Outcome Counts")
plt.savefig("plots/outcome_counts.png")
plt.close()

# BMI vs Category
sns.boxplot(x="BMI_Category", y="BMI", data=df)
plt.title("BMI by Category")
plt.savefig("plots/bmi_by_category.png")
plt.close()

# Correlation heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.savefig("plots/correlation_matrix.png")
plt.close()
```
