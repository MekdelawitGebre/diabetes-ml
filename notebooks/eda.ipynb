```python
# ===============================
# Diabetes EDA Notebook
# ===============================

import sys
from pathlib import Path
import warnings

# -------------------------------
# Project root setup
# -------------------------------
# Adjust if your notebook is inside a subfolder (e.g., 'notebooks/')
project_root = Path().resolve().parent
sys.path.append(str(project_root))

# -------------------------------
# Suppress pandas chained assignment warnings for cleaner output
# -------------------------------
warnings.simplefilter(action="ignore", category=FutureWarning)

# -------------------------------
# Imports
# -------------------------------
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

from src.data.preprocess import load_data, clean_data
from src.data.feature_engineering import add_age_bins, add_bmi_features

# -------------------------------
# Create plots directory
# -------------------------------
plots_dir = project_root / "plots"
plots_dir.mkdir(exist_ok=True)

# -------------------------------
# Load and preprocess data
# -------------------------------
df = load_data()
df = clean_data(df)
df = add_age_bins(df)
df = add_bmi_features(df)

# -------------------------------
# Plot 1: Glucose Distribution
# -------------------------------
plt.figure(figsize=(8,6))
sns.histplot(df["Glucose"], bins=30, kde=True)
plt.title("Glucose Distribution")
plt.xlabel("Glucose")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(plots_dir / "glucose_distribution.png")
plt.close()

# -------------------------------
# Plot 2: Outcome Counts
# -------------------------------
plt.figure(figsize=(6,5))
sns.countplot(x="Outcome", data=df)
plt.title("Diabetes Outcome Counts")
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(plots_dir / "outcome_counts.png")
plt.close()

# -------------------------------
# Plot 3: BMI vs Category
# -------------------------------
plt.figure(figsize=(8,6))
sns.boxplot(x="BMI_Category", y="BMI", data=df)
plt.title("BMI by Category")
plt.xlabel("BMI Category")
plt.ylabel("BMI")
plt.tight_layout()
plt.savefig(plots_dir / "bmi_by_category.png")
plt.close()

# -------------------------------
# Plot 4: Correlation Heatmap (numeric features only)
# -------------------------------
numeric_df = df.select_dtypes(include="number")  # Only numeric columns

plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix (Numeric Features Only)")
plt.tight_layout()
plt.savefig(plots_dir / "correlation_matrix.png")
plt.close()

print(f"All plots saved in: {plots_dir}")
```
